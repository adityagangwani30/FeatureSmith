# Featuresmith Tutorial: Exploring Datasets & Statistical Profiling

In this guide, we will dive deep into Featuresmith's **Profiling Engine** using the **Sales** dataset (which contains numerical transactions, category labels, text descriptions, date times, and missing fields).

---

```python
import os

import featuresmith as fs

dataset_path = os.path.join("..", "data", "processed", "sales.csv")
dataset = fs.load(dataset_path)
```

### Step 1: Run Detailed Profiling

Let's compute statistics and retrieve the `ProfileResult` object.

```python
profile = fs.profile(dataset)
```

### Step 2: Exploring Numeric Profiles

Let's analyze numerical statistics for `sales_amount`.

```python
sales_prof = profile.column_profiles["sales_amount"]
stats = sales_prof.numeric_stats

if stats is not None:
    print("Sales Amount Numerical Profile:")
    print(f"- Mean   : {stats.mean:.2f}")
    print(f"- Median : {stats.median:.2f}")
    print(f"- Min    : {stats.min:.2f}")
    print(f"- Max    : {stats.max:.2f}")
    print(f"- Std Dev: {stats.std:.2f}")
    print(f"- Skew   : {stats.skewness:.4f}")
    print(f"- Kurt   : {stats.kurtosis:.4f}")
```

### Step 3: Exploring Categorical Profiles

Let's check the categories of product sales.

```python
cat_prof = profile.column_profiles["category"]
print("Category Profile:")
print(f"- Cardinality: {cat_prof.unique_count}")
print(f"- Logical Type: {cat_prof.logical_type}")
print("- Value frequencies:")
for val, count in (
    cat_prof.value_counts.items() if hasattr(cat_prof, "value_counts") else []
):
    print(f"  {val}: {count} orders")
```

### Step 4: Datetime Inferences

Featuresmith parses string datetimes and extracts min/max timestamps and the range of transactions.

```python
date_prof = profile.column_profiles["order_date"]
print("Order Date Profile:")
print(f"- Logical Type: {date_prof.logical_type}")
if hasattr(date_prof, "datetime_stats") and date_prof.datetime_stats is not None:
    dt = date_prof.datetime_stats
    print(f"- Min Date: {dt.min_val}")
    print(f"- Max Date: {dt.max_val}")
    print(f"- Span (days): {dt.range_days}")
```

### Step 5: Pearson Correlation Analysis

Featuresmith compiles Pearson correlations to map feature dependencies. Let's inspect them.

```python
print("Correlation matrix preview:")
for key, val in list(profile.correlations.items())[:5]:
    print(f"- {key}: Pearson correlation = {val:.4f}")
```
